```python
import pandas as pd

url = "https://davidmegginson.github.io/ourairports-data/airports.csv"
df = pd.read_csv(url)

# تصفية مطارات المغرب فقط
df_ma = df[df["iso_country"] == "MA"].copy()

# اختيار الأعمدة المفيدة لبُعد المطار
cols = [
    "id",
    "ident",
    "type",
    "name",
    "municipality",
    "iso_region",
    "iata_code",
    "gps_code",
    "latitude_deg",
    "longitude_deg",
]
df_ma = df_ma[cols]

# حفظ النتيجة في ملف جديد
df_ma.to_csv("morocco_airports.csv", index=False, encoding="utf-8-sig")

print("عدد المطارات المستخرجة:", len(df_ma))
print(df_ma.head(10))
```

```python
from pathlib import Path
import pandas as pd

onda_dir = Path("onda_excel")
files = sorted(list(onda_dir.glob("*.xls")) + list(onda_dir.glob("*.xlsx")))

print(f"عدد ملفات Excel في onda_excel: {len(files)}\n")

for f in files:
    try:
        xl = pd.ExcelFile(f)
        print(f"📄 {f.name}  -->  Sheets: {xl.sheet_names}")
    except Exception as e:
        print(f"⚠️ {f.name}  -->  خطأ في القراءة: {e}")
```
